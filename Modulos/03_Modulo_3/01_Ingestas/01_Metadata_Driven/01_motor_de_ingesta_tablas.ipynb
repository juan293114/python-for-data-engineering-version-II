```python
# Crear el widget
dbutils.widgets.text("tabla", "playing_with_neon", "Nombre de la tabla")

# Leer el valor
tabla = dbutils.widgets.get("tabla")

print(f"Tabla seleccionada: {tabla}")
```

```python

jdbc_url = f"jdbc:postgresql://{POSTGRES_HOST}:{POSTGRES_PORT}/{POSTGRES_DB}?sslmode=require"

query = f"SELECT * FROM ventas_oltp.{tabla}"

df = (spark.read
      .format("jdbc")
      .option("url", jdbc_url)
      .option("query", query)
      .option("user", POSTGRES_USER)
      .option("password", POSTGRES_PASSWORD)
      .option("driver", "org.postgresql.Driver")
      .load())
```

```python
schema = "bronze"

nombre_completo = f"workspace.{schema}.{tabla}"
df.write.mode("overwrite").option("mergeSchema", "true").format("delta").saveAsTable(nombre_completo)
```
