```python
# 03_silver/3_silver_transform.py
# SILVER: Enriquecimiento de datos

print("="*50)
print("SILVER: Enriquecimiento de datos")
print("="*50)

df_bronze = spark.table("capa_bronze.coffee_sales_bronze")

from pyspark.sql.functions import col, when, trim, date_format, hour

df_silver = df_bronze \
    .dropDuplicates(["Date", "Time"]) \
    .withColumn("coffee_name", trim(col("coffee_name"))) \
    .withColumn("cash_type", trim(col("cash_type"))) \
    .withColumn("Time_of_Day", trim(col("Time_of_Day"))) \
    .withColumn("Year", date_format(col("Date"), "yyyy")) \
    .withColumn("Month_Number", date_format(col("Date"), "MM")) \
    .withColumn("Day", date_format(col("Date"), "dd")) \
    .withColumn("Day_Of_Week", date_format(col("Date"), "EEEE")) \
    .withColumn("Month_Name_Full", date_format(col("Date"), "MMMM")) \
    .withColumn("Hour_Exact", hour(col("Time"))) \
    .withColumn("price_category",
                when(col("money") < 25, "Barato")
                .when(col("money") < 35, "Medio")
                .otherwise("Caro")) \
    .withColumn("is_weekend",
                when(col("Weekday").isin(["Sat", "Sun"]), "Weekend")
                .otherwise("Weekday"))

# Guardar en Silver
df_silver.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("capa_silver.coffee_sales_silver")

print(f" {df_silver.count()} registros en Silver")
print("="*50)
print(" SILVER COMPLETADO")
print("="*50)
```
