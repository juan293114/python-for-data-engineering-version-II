```python
# 01_raw/1_ingest_to_raw.py
#  INGESTA A RAW

print("="*50)
print(" INGESTA A RAW")
print("="*50)

# Crear esquemas
spark.sql("CREATE SCHEMA IF NOT EXISTS capa_raw")
spark.sql("CREATE SCHEMA IF NOT EXISTS capa_bronze")
spark.sql("CREATE SCHEMA IF NOT EXISTS capa_silver")
spark.sql("CREATE SCHEMA IF NOT EXISTS capa_gold")
print(" Esquemas creados")

# Leer CSV
df_raw = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .table("coffee_sales_raw")

# Guardar en RAW
df_raw.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("capa_raw.coffee_sales_raw")

print(f" {df_raw.count()} registros en RAW")
display(df_raw.limit(5))

print("="*50)
print(" RAW COMPLETADO")
print("="*50)
```
