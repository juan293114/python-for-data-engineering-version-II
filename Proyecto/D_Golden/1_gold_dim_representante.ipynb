```python
spark.sql("DROP TABLE IF EXISTS workspace.gold.dim_representante")
```

```python
spark.sql(
    """
    create table if not exists workspace.gold.dim_representante(
        id_representante bigint,
        representante string,
        ciudad string
    )
   
    """
)
```

```python
from pyspark.sql import functions as F
df_dim_representante=spark.table("workspace.silver.tbl_representantes")
df_dim_representante=(df_dim_representante
                      .dropDuplicates(["representante","ciudad"])
                      .withColumn("id_representante",F.monotonically_increasing_id()+1)
                      .select("id_representante","representante","ciudad")
                      )
```

```python
df_dim_representante.write.format("delta")\
    .option("mergeShema","true")\
    .mode("overwrite")\
    .saveAsTable("workspace.gold.dim_representante")
```
