```python
def get_silver_config(entity_name: str) -> dict:
    df = spark.sql(f"""
        SELECT *
        FROM workspace.control.silver_transform_config
        WHERE entity_name = '{entity_name}' AND is_active = true
    """)
    filas = df.collect()
    if len(filas) == 0:
        raise ValueError(f"No existe configuración Silver activa para '{entity_name}'")
    return filas[0].asDict()
```
