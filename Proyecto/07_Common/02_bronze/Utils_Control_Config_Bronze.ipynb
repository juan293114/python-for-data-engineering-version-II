```python
def get_bronze_config(entity_name: str) -> dict:
    """Lee la fila de configuración de destino en Bronze para esta entidad,
    incluyendo su load_mode, watermark_column y last_watermark."""
    df = spark.sql(f"""
        SELECT *
        FROM workspace.control.bronze_load_config
        WHERE entity_name = '{entity_name}' AND is_active = true
    """)
    filas = df.collect()
    if len(filas) == 0:
        raise ValueError(f"No existe configuración Bronze activa para '{entity_name}'")
    return filas[0].asDict()


def get_raw_path_for_entity(entity_name: str) -> str:
    """Consulta puntual a raw_ingestion_config, solo para obtener la ruta física."""
    df = spark.sql(f"""
        SELECT raw_path
        FROM workspace.control.raw_ingestion_config
        WHERE entity_name = '{entity_name}' AND is_active = true
    """)
    filas = df.collect()
    if len(filas) == 0:
        raise ValueError(f"No existe configuración Raw activa para '{entity_name}'")
    return filas[0].raw_path
```
