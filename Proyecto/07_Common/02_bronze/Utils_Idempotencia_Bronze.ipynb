```python
def bronze_already_loaded_today(target_table: str, execution_date: date = None) -> bool:
    execution_date = execution_date or date.today()
    try:
        conteo = spark.sql(f"""
            SELECT COUNT(*) AS total
            FROM {target_table}
            WHERE _load_date = '{execution_date}'
        """).collect()[0].total
        return conteo > 0
    except Exception:
        return False
```
