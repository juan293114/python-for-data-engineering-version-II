```python
def build_raw_file_name(entity_name: str, execution_date: date = None) -> str:
    execution_date = execution_date or date.today()
    fecha_str = execution_date.strftime("%Y_%m_%d")
    return f"{entity_name}_{fecha_str}.json"
```

```python
def path_exists(path: str) -> bool:
    """Helper genérico: ¿existe este archivo/carpeta en el Volumen?"""
    try:
        dbutils.fs.ls(path)
        return True
    except Exception:
        return False
```
