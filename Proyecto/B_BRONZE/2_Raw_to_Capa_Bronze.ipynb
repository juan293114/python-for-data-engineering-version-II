```python
from pathlib import Path
import pandas as pd
raw_root = Path("/Volumes/internet_fijo_elt/raw/archivos_internet_fijo/")

print(raw_root)
```

```python
candidate_files = sorted(
    raw_root.glob("*.xlsx")
)

print(f"Archivos encontrados: {len(candidate_files)}")

for file in candidate_files:
    print(file)
```

```python
%pip install openpyxl

excel_file = candidate_files[0]

df_pandas = pd.read_excel(
    excel_file,
    sheet_name="Dataset"
)

print(f"Registros: {len(df_pandas)}")
print(f"Columnas: {len(df_pandas.columns)}")

display(df_pandas.head())
```

```python
df_spark = spark.createDataFrame(df_pandas)

df_spark.printSchema()

display(df_spark.limit(10))
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS internet_fijo_elt.bronze
COMMENT 'Capa Bronze: datos crudos procesados';
```

```python
bronze_table = (
    "internet_fijo_elt.bronze.conexiones_internet_fijo"
)

df_spark.write \
    .format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable(bronze_table)

print(f"Tabla Bronze creada: {bronze_table}")
```

```python
bronze_df = spark.table(
    "internet_fijo_elt.bronze.conexiones_internet_fijo"
)

print(f"Registros Bronze: {bronze_df.count()}")

display(bronze_df.limit(20))
```
