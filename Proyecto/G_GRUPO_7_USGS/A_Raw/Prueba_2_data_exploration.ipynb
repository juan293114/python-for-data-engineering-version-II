```python
import json

raw_peru = "/Volumes/workspace/raw/usgs_earthquakes/peru/2026/07/earthquakes_2026-07-01_2026-08-01.json"
raw_chile = "/Volumes/workspace/raw/usgs_earthquakes/chile/2026/07/earthquakes_2026-07-01_2026-08-01.json"

with open(raw_peru, "r", encoding="utf-8") as file:
    data_peru = json.load(file)

print(type(data_peru))
print(data_peru.keys())

print(data_peru["metadata"])

print("Cantidad de eventos: ", len(data_peru["features"]))

```

```python
primer_evento = data_peru["features"][0]
print("ID:", primer_evento["id"])
print("Propiedades:", primer_evento["properties"])
print("Geometría:", primer_evento["geometry"])
```

```python
#Revisamos los campos disponibles en los eventos

print("Campos principales:")
print(primer_evento.keys())

print("Campos de properties:")
print(primer_evento["properties"].keys())

print("Campos de geometry:")
print(primer_evento["geometry"].keys())
```

```python
#Revisamos valores nullos por campo

campos = primer_evento["properties"].keys()

for campo in campos:
    cantidad_nulos = sum(
        1 for evento in data_peru["features"]
        if evento["properties"].get(campo) is None
    )

    print(f"{campo}: {cantidad_nulos}")
```

```python
#Cargamos y revisamos valores nullos de Chile

with open(raw_chile, "r", encoding="utf-8") as file:
    data_chile = json.load(file)

print("cantidad de eventos en Chile:", len(data_chile["features"]))

primer_evento_chile = data_chile["features"][0]

campos_chile = primer_evento_chile["properties"].keys()

for campo in campos_chile:
    cantidad_nulos = sum(
        1 for evento in data_chile["features"]
        if evento["properties"].get(campo) is None
    )

    print(f"{campo}: {cantidad_nulos}")
```
