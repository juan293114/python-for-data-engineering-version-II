```python
from pyspark.sql.functions import *

silver_df = spark.read.table(
    "workspace.silver.earthquakes"
)
#display(silver_df)
```

```python
dim_magnitude = spark.createDataFrame(
    [
        (1, 0, 4, "Bajo"),
        (2, 4, 5, "Moderado"),
        (3, 5, 6, "Fuerte"),
        (4, 6, 15, "Severo")
    ],
    [
        "magnitude_key",
        "min_magnitude",
        "max_magnitude",
        "category"
    ]
)

display(dim_magnitude)
```

```python
(
    dim_magnitude.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        "workspace.gold.dim_magnitude"
    )
)

print("dim_magnitude creada")
```
