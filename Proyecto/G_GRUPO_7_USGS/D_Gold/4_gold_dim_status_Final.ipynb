```python
from pyspark.sql.functions import *

silver_df = spark.read.table(
    "workspace.silver.earthquakes"
)
#display(silver_df)
```

```python
dim_status = (
    silver_df
    .select(
        col("status")
    )
    .dropDuplicates()
)
```

```python
from pyspark.sql.functions import monotonically_increasing_id

dim_status = (
    dim_status
    .withColumn(
        "status_key",
        monotonically_increasing_id()
    )
)
```

```python
dim_status = dim_status.select(
    "status_key",
    "status"
)
display(dim_status)
```

```python
(
    dim_status.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        "workspace.gold.dim_status"
    )
)

print("dim_status creada")
```
