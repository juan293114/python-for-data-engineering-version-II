```python
silver_df = spark.read.table(
    "workspace.silver.earthquakes"
)
#display(silver_df)
```

```python
from pyspark.sql.functions import *

dim_date = (
    silver_df
    .select(
        date_format(
            col("event_datetime"),
            "yyyyMMdd"
        )
        .cast("int")
        .alias("date_key"),

        to_date(
            col("event_datetime")
        )
        .alias("date"),

        year(
            col("event_datetime")
        )
        .alias("year"),

        month(
            col("event_datetime")
        )
        .alias("month"),

        date_format(
            col("event_datetime"),
            "MMMM"
        )
        .alias("month_name"),

        quarter(
            col("event_datetime")
        )
        .alias("quarter")
    )
    .dropDuplicates()
)

#display(dim_date)
```

```python
(
    dim_date.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        "workspace.gold.dim_date"
    )
)

print("dim_date creada")
```
