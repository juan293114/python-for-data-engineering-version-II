#Catalogo

```python
#Catalogo Actual
catalogo = spark.sql(f""" SELECT CURRENT_CATALOG()""").first()[0]
```

```python
schema = "raw"
```

#Volumenes

```python
volumes_config = {
    "products": "Archivos json  estado puro tal como se extraen del origen",
    "carts": "Archivos json  estado puro tal como se extraen del origen",
    "users": "Archivos json  estado puro tal como se extraen del origen",
}
# DBTITLE 1,Creando schema
for volume, comment in volumes_config.items():
    try:
        spark.sql(f"""
                  CREATE VOLUME IF NOT EXISTS {catalogo}.{schema}.{volume}
                  COMMENT '{comment}'
                  """)
        print(f"Volume {volume} creado exitosamente")
    except Exception as e:
        print(f"Error al crear el volume {volume}: {e}")
```
