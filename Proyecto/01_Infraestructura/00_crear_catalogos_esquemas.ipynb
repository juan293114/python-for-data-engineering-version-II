#Catálogos

```python
#Seteamos en qué schema estamos

catalogo = spark.sql("SELECT CURRENT_CATALOG()").collect()[0][0]
```

#Schemas

```python
#Adicionar Schema si es que se necesita alguno
schemas_config = {
    "raw": "Datos en estado puro tal como se extraen del origen.",
    "bronze": "Datos crudos tabulares con metadatos de ingesta.",
    "silver": "Datos limpios, validados y estructurados.",
    "control": "Metadata operativa del pipeline."
}

#Creación de schemas
for schema, comment in schemas_config.items():
    try:
        spark.sql(f"""
                  CREATE SCHEMA IF NOT EXISTS {catalogo}.{schema}
                  COMMENT '{comment}'
                  """)
        print(f"Schema {schema} creado exitosamente")
    except Exception as e:
        print(f"Error al crear el schema {schema}: {e}")
```
