```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```python
spark.sql("SHOW SCHEMAS IN internet_fijo_elt").show()
```

```python
df = spark.table("internet_fijo_elt.silver.conexiones_internet_fijo")

dim_empresa = (
    df
    .select("empresa")
    .distinct()
    .withColumn("id_empresa", F.row_number().over(Window.orderBy("empresa")))
    .select("id_empresa", F.col("empresa").alias("nombre_empresa"))
)
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS internet_fijo_elt.gold.dim_empresa (
 id_empresa INT,
 nombre_empresa STRING
)
""")

dim_empresa.write.mode("append").saveAsTable("internet_fijo_elt.gold.dim_empresa")
```
