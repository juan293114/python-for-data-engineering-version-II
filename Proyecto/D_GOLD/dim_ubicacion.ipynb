```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

df = spark.table("internet_fijo_elt.silver.conexiones_internet_fijo")

dim_ubicacion = (
    df
    .select("departamento", "provincia", "distrito")
    .distinct()
    .withColumn("id_ubicacion", F.row_number().over(Window.orderBy("departamento", "provincia", "distrito")))
    .select("id_ubicacion", "departamento", "provincia", "distrito")
)
```

```python
dim_ubicacion.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("internet_fijo_elt.gold.dim_ubicacion")
```
