```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

df = spark.table("internet_fijo_elt.silver.conexiones_internet_fijo")

dim_tecnologia = (
    df
    .select("tecnologia")
    .distinct()
    .withColumn("id_tecnologia", F.row_number().over(Window.orderBy("tecnologia")))
    .select("id_tecnologia", F.col("tecnologia").alias("tecnologia"))
)
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS internet_fijo_elt.gold.dim_tecnologia (
 id_tecnologia INT,
 tecnologia STRING
)
""")

dim_tecnologia.write.mode("append").saveAsTable("internet_fijo_elt.gold.dim_tecnologia")
```
