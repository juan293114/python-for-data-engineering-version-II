```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

df_silver = spark.table("internet_fijo_elt.silver.conexiones_internet_fijo")

display(df_silver.limit(10))
```

```python
print("Registros Silver:", df_silver.count())
```

```python
spark.sql("SHOW SCHEMAS IN internet_fijo_elt").show()
```

```python
dim_empresa = (
    df_silver
    .select("empresa")
    .distinct()
    .withColumn(
        "empresa_id",
        F.row_number().over(
            Window.orderBy("empresa")
        )
    )
)
```

```python
dim_empresa_renamed = dim_empresa.select(
    F.col("empresa_id").alias("id_empresa"),
    F.col("empresa").alias("nombre_empresa")
)

dim_empresa_renamed.write.format("delta").mode("overwrite").saveAsTable(
    "internet_fijo_elt.gold.dim_empresa"
)
```
