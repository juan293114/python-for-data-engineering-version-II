```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

df_silver = spark.table("internet_fijo_elt.silver.conexiones_internet_fijo")

display(df_silver.limit(10))
```

```python
dim_segmento = (
    df_silver
    .select("segmento")
    .distinct()
    .withColumn(
        "segmento_id",
        F.row_number().over(
            Window.orderBy("segmento")
        )
    )
)
```

```python
dim_segmento_renamed = dim_segmento.select(
    F.col("segmento_id").alias("id_segmento"),
    F.col("segmento")
)

dim_segmento_renamed.write.format("delta").mode("overwrite").saveAsTable(
    "internet_fijo_elt.gold.dim_segmento"
)
```
