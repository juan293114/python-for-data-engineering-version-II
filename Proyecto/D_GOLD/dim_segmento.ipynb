```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

df = spark.table("internet_fijo_elt.silver.conexiones_internet_fijo")

dim_segmento = (
    df
    .select("segmento")
    .distinct()
    .withColumn("id_segmento", F.row_number().over(Window.orderBy("segmento")))
    .select("id_segmento", F.col("segmento").alias("segmento"))
)
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS internet_fijo_elt.gold.dim_segmento (
 id_segmento INT,
 segmento STRING
)
""")

dim_segmento.write.mode("append").saveAsTable("internet_fijo_elt.gold.dim_segmento")
```
