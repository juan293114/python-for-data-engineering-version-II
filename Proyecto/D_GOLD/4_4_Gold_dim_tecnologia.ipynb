```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

df_silver = spark.table("internet_fijo_elt.silver.conexiones_internet_fijo")

display(df_silver.limit(10))
```

```python
dim_tecnologia = (
    df_silver
    .select("tecnologia")
    .distinct()
    .withColumn(
        "tecnologia_id",
        F.row_number().over(
            Window.orderBy("tecnologia")
        )
    )
)
```

```python
dim_tecnologia_renamed = dim_tecnologia.select(
    F.col("tecnologia_id").alias("id_tecnologia"),
    F.col("tecnologia")
)

dim_tecnologia_renamed.write.format("delta").mode("overwrite").saveAsTable(
    "internet_fijo_elt.gold.dim_tecnologia"
)
```

```python
print("Dim Tecnología:", dim_tecnologia.count())
```
