```python
spark.sql("""
DROP TABLE IF EXISTS workspace.weather_gold.dim_weather_region
""")
```

```python
# 1. Crear la tabla workspace.weather_gold.dim_weather_region
spark.sql("""
CREATE TABLE IF NOT EXISTS workspace.weather_gold.dim_weather_region (
  weather_region_id INT,
  weather_region STRING,
  description STRING
)
""")
```

```python
# 2. Poblar de data la dimension
dim_weather_region = spark.createDataFrame(
    [
        (1, "Costa", "Región costera del Perú"),
        (2, "Sierra", "Región andina del Perú"),
        (3, "Selva", "Región amazónica del Perú")
    ],
    [
        "weather_region_id",
        "weather_region",
        "description"
    ]
)

dim_weather_region.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.weather_gold.dim_weather_region")
```
